## Description du dataset

### Mesures de polluants (avec capteurs de référence GT et capteurs PT08)
* CO(GT) - Concentration en monoxyde de carbone mesurée par l'analyseur de référence (en mg/m³)
* PT08.S1(CO) - Réponse du capteur semi-conducteur PT08.S1 pour le CO (valeurs normalisées)
* NMHC(GT) - Concentration en hydrocarbures non méthaniques mesurée par l'analyseur de référence (en µg/m³)
* C6H6(GT) - Concentration en benzène mesurée par l'analyseur de référence (en µg/m³)
* PT08.S2(NMHC) - Réponse du capteur semi-conducteur PT08.S2 pour les NMHC (valeurs normalisées)
* NOx(GT) - Concentration en oxydes d'azote mesurée par l'analyseur de référence (en ppb)
* PT08.S3(NOx) - Réponse du capteur semi-conducteur PT08.S3 pour les NOx (valeurs normalisées)
* NO2(GT) - Concentration en dioxyde d'azote mesurée par l'analyseur de référence (en µg/m³)
* PT08.S4(NO2) - Réponse du capteur semi-conducteur PT08.S4 pour le NO2 (valeurs normalisées)
* PT08.S5(O3) - Réponse du capteur semi-conducteur PT08.S5 pour l'ozone (valeurs normalisées)
* Conditions météorologiques
* T - Température de l'air (probablement en °C)
* RH - Humidité relative (en pourcentage, 0-100%)
* AH - Humidité absolue (probablement en g/m³)
* DateTime - Horodatage des mesures

```python
# Remplacer les -200.0 (ce que je pense: -200 est une valeur pour marquer qu'il n ya pas de donnée enrefistrée comme si on a -1 pour dire non dispo)
df_clean = df_imputed.replace(-200.0, np.nan)
```

```python
df_clean.describe()
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')
df_clean[numeric_cols] = imputer.fit_transform(df_clean[numeric_cols])
```

```python
df_clean.isnull().sum()
```

```python
def remove_outliers_iqr(df, cols):
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
    return df

df_no_outliers = remove_outliers_iqr(df_clean.copy(), numeric_cols)
```
